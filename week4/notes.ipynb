## Function delcarations
we can create custom functions that do specific tasks in python. I'll use the above running time n seconds conversion as an example.

```python
# this fucntion takes a string formatted like "30:15" and returns an int representing total times in seconds
def calculate_seconds(time_string): # def + function name+arguments in parentheses + colon
# we are expecting a string oformatted like "30:15" as input
# which will represented by a placeholder "time_string"
    minutes,seconds = time_string.split(":")
    total_seconds= int(minutes)*60 + int(seconds) # converting + calculating total seconds
    return total_seconds

print(calculate_seconds("40:45"))
```

function declarations are made of the following elements:
- the `def` keyword: tells python that we are defining a new function!
- the name of the new function (`calculate_seconds`)
- the arguments: multiple arguments aare separated by commas
- the arguments are then processed by the lines of code that are indented below the "def" line
- `return` specifies what data is output by the function

```python
def angry(text):
    text = text.upper()
    exclamations="!"*len(text)
    return text + exclamations

print(angry("hello world"))
```

## Conditionals
conditionals, like loops are another fundamental structure found across most programming languages
conditionals rely in comparison operators, which compare two values and return a value of `true` or `false`
- `x == y` equals
- `x != y` not equal
- `x < y` less than
- `x > y` greater than
- `x <= y` less than or equal
- `x >= y` greater than or equal

comparison operators can also be combined with logical operators which allow you to check more than ome ting at once (or, in this case of the not operator, the opposite of a true/false value)

- `x > y and a==b` are BOTH conditions true?
- `x < y or a > b` are at least one of these conditions true?


```python
x=5
y=6
a=1
b=1

print(x>y)
print(x<y)
print(x==y)
print(x<y and a!=b)
print(x>y or a>b)
```

```python
temperature = 45

if temperature<60:
    print("put on a jacket")
else:
    print("no jacket needed!")
```

In the above simple example, we see the following elements:

- if `if` keyword, which initializes theconditional statement
- the comparison (temperature <60), which is evaliated as true or false
- if the comparison evaluates to `true`, the indented portion directly below the if statement runs
- if the comparison is not `true`, the indented portion below `else:` runs

We can also stack conditions using the else if or `elif` keyword:

```python
temperature = 30

if temperature >= 80:
    print("wear shorts")
elif temperature >= 65:
    print("wear pants")
elif temperature >= 50:
    print("wear a sweater")
elif temperature >= 35:
    print("Wear a jacket")
else:
    print("stay home!!")
```

We can add logical operators (like `and` & `or`) to combine comparisons in conditionals:

```python
temeprture =80

if temperature >=65 and temperature <= 75:
    print("this is comfortable")
elif temperature <32 or temperature<95:
    print("the weather is extreme")
else:
    print("meh")
```

```python
# and here is the robust solution for large datasets

love_songs = [] #this is an empty list! we can add stuff to it

for album in beatles_albums: # for each album in the set...
    for track in album["tracks"]: # for each track in each album ...
        if " lov" in (" "+track).lower(): # artificially adding a training space to the beginning of the tracklist title to include all results.
            love_songs.append(track)

print(love_songs)
print(len(love_songs))
```

## working with CSV data

CSV is one of the most common dataset formats, its basically a spreadsheet with rows and columns
each row is a new line, each column is separated by a comma
CSVs usually have a "header" row that is in the first row of the docuemnt
each subsequent row is the actual data!

to work with CSV data in python, we need to import the CSV library by typing `import CSV` before attempting to use any CSV functions. In this case, CSV is in the same folder as the workbook

we cna open a file using the following syntax:

`with open(file_name, options) as file:`

```python
import csv

with open("moma_artists.csv", "r", encoding="utf-8") as file: # 'r' means 'read' referring to it from now on as 'file'
    reader = csv.reader(file) # initializing an object that interprets CSV data
    #the below line will extract all the headings!
    header = next(reader) # telling the reader to read the first line and move on to the next
    rows = list(reader)   # takes every other row in the dataser and makes it into a list

print(header)
print(rows[0]) # prints the first row
```

we could enumerate a number with the values to make crossreferencing to the headers easier. (week 7)

We can also generate a list of dictionaries that use the header to establish a set of shares keys, assigning values for each row based on those keys/headings

we can do that using the csv mehtod `DictReader(file)`

```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) # this will interpret the csv as a list of dictionaries
    data = list(reader) # stores the CSV data as a list of dictionaries we can access later

print(data[-1]) # print last thing in dataset
```

```python
# fulter by nationality: only Swedish artists

swedish_artists = []

for artist in data: 
    if artist["Nationality"] == "Swedish":
        swedish_artists.append(artist) # it will append all the biogrphical data from each list containng nationality=swedish to the new list

print(len(swedish_artists))
print(swedish_artists[:10])

for artist in swedish_artists[:10]: # if we only want names
    print(artist["DisplayName"])
    
```

```python
young_artist = []
for artist in data:
    if int(artist["BeginDate"]) > 1990:
        young_artist.append(artist)

print(len(young_artist))

for artist in young_artist[:10]:
    print(artist["DisplayName"])
```
