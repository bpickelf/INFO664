```python
with open("moma_artists.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file) 
    data = list(reader) 

print(len(data))
print(data[-1]) 
```

```python
artists_with_gender_data = []

for artist in data:
    if artist["Gender"] is not None and artist["Gender"] != "":
        artists_with_gender_data.append(artist)

print(len(artists_with_gender_data))

for artist in artists_with_gender_data[:10]:
    print(artist["DisplayName"])
```

```python
female_artists = []

for artist in data:
    if artist["Gender"] == "female" or "woman" in artist["Gender"]:
        female_artists.append(artist)

print(len(female_artists))

for artist in female_artists[:10]:
    print(artist["DisplayName"])
```

```python
male_artist = []
for artist in data:
    if artist["Gender"]== "male" or " man" in artist["Gender"]:
        male_artist.append(artist)

print(len(male_artist))

for artist in male_artist[:10]:
    print(artist["DisplayName"],artist["Gender"])
```

```python
non_binary_artists = []

for artist in data:
    if artist["Gender"]== "non-binary":
       non_binary_artists.append(artist)

print(len(non_binary_artists))

for artist in non_binary_artists[:10]:
    print(artist["DisplayName"],)
```

```python
other_artists = []

for artist in data:
    if artist["Gender"]!= "non-binary" and artist["Gender"]!= "male" and artist["Gender"] != "female" and ["Gender"] is not None and artist["Gender"] != "" and "woman" not in artist["Gender"] and " man" not in artist["Gender"]:
       other_artists.append(artist)

print(len(other_artists))

for artist in other_artists[:10]:
    print(artist["DisplayName"], artist["Gender"])
```
