```python
import csv #brings in all the csv-related functions

# the open function takes all 3 arguments:
# - the file name/path
# - "r" for read
# - and the encoding type

with open ("moma_artists.csv", "r", encoding= "utf-8") as file:
    reader = csv.DictReader(file) #read the file as dictionary data
    data = list(reader) # covert that dictionary object into a usable list called "data"

print(data[0]) # prints the first dictionary item 
```

```python
# filter out years with 0

artists_with_year = []

for artist in data:
    if artist['BeginDate'] != "0": # Add everything that is NOT a zero
        artists_with_year.append(artist)

print(len(data))
print(len(artists_with_year))
```

```python
artists_by_year_filtered = sorted(artists_with_year, key=lambda artist: int(artist['BeginDate']))

print(artists_by_year_filtered[0])
print(artists_by_year_filtered[0])
```

```python
for artist in artists_by_year_filtered[-20:]:
    print(artist['DisplayName'], artist['BeginDate'])
```

```python
unique_years = list(set(row["BeginDate"] for row in artists_by_year_filtered))

print(unique_years)
```

```python
sorted_unique_years=sorted(unique_years)

print(sorted_unique_years)
```

```python
count_by_year = {year: 0 for year in unique_years}

for artist in artists_by_year_filtered:
    year = artist["BeginDate"]
    count_by_year[year] += 1
```

```python
print(count_by_year)
```

```python
sorted_count_by_year=sorted(count_by_year)
```

```python
for year in (sorted_count_by_year)[:10]:
    print(year, count_by_year[year])
```

```python
for year in (sorted_count_by_year)[-10:]:
    print(year, count_by_year[year])
```

```python
# Professor's SOLUTION from class

year_counts={} # empty dictionary to store year counts

for artist in artists_by_year_filtered: # go through each artist in the filtered lsit
    year=artist['BeginDate'] #store the begin sate for each artist
    if year in year_counts: #check to see if that begindate is in the year
        year_counts[year] = year_counts[year]+1 #If true, add 1 to existing count
    else:
        year_counts[year]=1 # if that year isn't in the dictionary, create a key for the year and set value to 1

print(year_counts['1956'])
```

```python
sorted_years=sorted(year_counts, key=lambda year: year_counts[year], reverse=True) # ordering the list from highest count to lowest

print(sorted_years[:10])
```
