Using the queries from week 6 to pull for artwork, geography, and make it a CSV

```python
# redo my artist search in homework

import requests

base_url = "https://collectionapi.metmuseum.org"
search_path = "/public/collection/v1.1/search"
search_query = "?q=Shawnee" # the %20 stops the URL from breaking
artist_or_culture = "&artistOrCulture=true" # flag that limits results to the artist or culture fields
limit = "&limit=500" # sets the amount of responses per page to 500
offset = "&offset=0"

url = f'{base_url}{search_path}{search_query}{artist_or_culture}{limit}{offset}'
search_response = requests.get(url)

print(url)
```

```python
print(search_response.ok)
```

```python
shawnee_objects = search_response.json()
complete_shawnee_objects = shawnee_objects["objectIDs"]
print(len(complete_shawnee_objects))
print ("out of")
print(shawnee_objects['total'])
```

```python
from time import sleep # this allows us to use the time finctions related to waiting between operations

object_path = "/public/collection/v1/objects/" # this is the path of the ibjects endpoint in the Met API
# antelope_item=antelope_object_ids[0] # this is the first item in the list of objectIDs above

# url = f"{base_url}{object_path}{antelope_item}"

# response_object = requests.get(url)

# print(response_object.ok)
# print(url)

all_shawnee_details=[]

for shawnee_artwork in complete_shawnee_objects: # for each of the object IDs I pulled from the MetAPI earlier... limiting to the first 10 as an example
    url = f"{base_url}{object_path}{shawnee_artwork}"
    response_object=requests.get(url) # making the request to the server
    all_shawnee_details.append(response_object.json())
    print(response_object.json()['artistDisplayName']) # prints the artist name to be sure our criteria is correct
    print(f"grabbed details from objectID {shawnee_artwork} with status code {response_object.status_code}")
    sleep(0.5) # wait half a second before moving on to the next step in the rate limiting in the Met's server

print(len(all_shawnee_details)) # if the item isn't found, and returning a 404, something is off. 

print(all_shawnee_details[0])
```

```python
import csv   # brings in all the csv-related functions

with open("all_shawnee_artworks.csv", "w", encoding="utf-8") as file: # initiate the opening of a file, and write a new file
    writer = csv.DictWriter(file, all_shawnee_details[0].keys()) # this creates the columns in the CSV file from the keys
    writer.writeheader() # Creates the first row of the CSV, containing the HEaders
    writer.writerows(all_shawnee_details) # goes through every item in the yoko_ono_artworks list and put it in the CSV file
    # you should see it in your folder after running.
```
