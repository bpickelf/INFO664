## Data types in Python
Values in Python can be a number of different data types, such as **int, float, string, and bool**

```python
print(type(5))   #INT
print(type(5.5)) #The innermost function will evaluate, then print function will run. 5.5 is a FLOAT (decimal-point number)
print(type("five")) #STR (string - text)
print(type("5")) #STR (in quotations so still a text)
print(type(True)) #BOOL (boolean - true or false)
```
