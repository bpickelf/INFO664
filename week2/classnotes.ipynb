## Data types in Python
Values in Python can be a number of different data types, such as **int, float, string, and bool**

```python
print(type(5))   #INT
print(type(5.5)) #The innermost function will evaluate, then print function will run. 5.5 is a FLOAT (decimal-point number)
print(type("five")) #STR (string - text)
print(type("5")) #STR (in quotations so still a text)
print(type(True)) #BOOL (boolean - true or false)
```

## Arithmitic operators

## Naming conventions with variables

- your variable name should represent what it is holding
- names **cannot** start with a number
- names **cannot** have a hyphen in them
- names **CAN** have a underscore in them
  - place_of_birth is a valid variable name using "Snake Case"
  - USE_UPPERCASE to designate a constant variable that will to change after its set
- Names **cannot** be special words in the python language like "if" and "for"

## Converting Data types
data can be converted from one type to another using conversion functions:

```python
pi = 3.14159
print(pi)
```

```python
pi_as_string = str(pi) # str function converts to a string!
print(pi_as_string)
print(type(pi_as_string))


pi_as_float_again= float(pi_as_string)
print(pi_as_float_again)
print(type(pi_as_float_again))
```

```python
print(pi_as_string>3) #will not work because they are not the same datatype
```

```python
print(pi_as_float_again>3) #will work becuase all numeric
```

## combining Strings (AKA concatenating)
strings can be combines in a number of ways. one way is to the *+* operator

```python
dog_name = "Nina"
dog_breed = "poodle"
dog_age = 12
my_dog_string= "My dog's name is " + dog_name + " and she is a " + str(dog_age) + " year old " + dog_breed
#must convert everything to the same datatype before concatenating
print(my_dog_string)
```

[F-strings](https://www.w3schools.com/python/python_strings_format.asp)

```python
# F-strings
my_dog_string=f"My dog's name is {dog_name} and she is a {dog_age} year old {dog_breed}." 
# this is a string that will having string literal formats that will insert those in with curly brackets
print(my_dog_string)
```

## Converting String Case
There are many different built in string methods for converting upper, lower, capitalized and title case.

```python
print(dog_name.upper())
print(dog_name.lower())
print(my_dog_string.title()) # not really representative of a true title case, so that will take extra work.
print(my_dog_string.capitalize())
```

## Searching, Counting, Extracting
There are some string methods in Python that deal with searching for substrings, etc. One of those is the keyword **in**.

```python
print("Cat" in my_dog_string) # this will tell us if that word 'cat' was found in the string
print("dog" in my_dog_string)
```

```python
my_dog_string=my_dog_string + " I love dogs and dog is man's best friend!"
print(my_dog_string) 
#if you run a cell multiple times, it will reprint everything or throw errors becuase it doesn't understan the variables you declared yet.
```

```python
dog_count=my_dog_string.count("dog")
print(f"The word 'dog' appears {dog_count} times.")
```
