## Saving CSVs

quick example for how to save a CSV from a list of dictionaries

```python
import csv   # brings in all the csv-related functions

with open("moma_art.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)   # read the file as dictionaries
    data = list(reader)             # convert that into a usable list called "data"

print(len(data))
print(data[0].keys())
```

```python
# filtering the large dataset into a smaller list of dictionaries with only works by Yoko Ono

yoko_ono_artworks = []

for artwork in data: 
    if artwork['Artist']== "Yoko Ono":
        yoko_ono_artworks.append(artwork)

print(len(yoko_ono_artworks))
```

```python
with open("yoko_ono_artworks.csv", "w", encoding="utf-8") as file: # initiate the opening of a file, and write a new file
    writer = csv.DictWriter(file, yoko_ono_artworks[0].keys()) # this creates the columns in the CSV file from the keys
    writer.writeheader() # Creates the first row of the CSV, containing the HEaders
    writer.writerows(yoko_ono_artworks) # goes through every item in the yoko_ono_artworks list and put it in the CSV file
    # you should see it in your folder after running.
    
```

## Regular Expressions

its own programming language to find a string of text that is structured within a larger body of text. Good for finding patterns in language to do complex searches. They use a bunch of specific special characters which represent different sequences

```python
caption = "William Wegman, Cotto, 1970. Gelatin silver print, 7 7/8 × 7 3/4 in. (20 × 19.7 cm). Whitney Museum of American Art, New York; purchase with funds from the Mrs. Percy Uris Purchase Fund and the Photography Committee 92.14 © William Wegman"
```

```python
import re # importing regular expressions functions/methods

match=re.search(r"([0-9]{4})", caption)

print(match)

year=match.group(1)

print(year)
```

```python
# open last week's moma artist data

with open ("../week5/moma_artists.csv", "r", encoding= "utf-8") as file: # goes back to the root folder, then to week 5, then to the file
    reader = csv.DictReader(file) #read the file as dictionary data
    artist_data = list(reader) # covert that dictionary object into a usable list called "data"

print(artist_data[0]) # prints the first dictionary item 
```

```python
catherines = []

for artist in artist_data:
    if re.search(r"^[CK]ath\W*", artist['DisplayName']):
        # searches for the beginning of the string, check if the firt character is a c or a k, followed by a "ath" followed by a sequence
        catherines.append(artist)

for catherine in catherines:
    print(catherine('DisplayName'))
```

## JSON

Unlike CSVs, JSON is not organized as a table. They are organized into objects that are nested lists! this is better for repeatable values. In CSVs repeated values have to be separated by commas in one cell (confusing and problematic!). They are interpreted as lists of dictionaries in Python.

```python
import json # import JSON methods

with open("unesco_cultural_sites.json", "r", encoding="utf-8") as file:
    sites = json.load(file)

print(len(sites))
print(type(sites[0]))
print(sites[0].keys())
print(sites[0]['name_en'])
```
