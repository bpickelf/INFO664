```python
# below is a method found using ChatGPT to try using loops to get all 263 results

import requests

url = "https://collectionapi.metmuseum.org/public/collection/v1.1/search"

# define parameters
params = {
    "q": "Monkey",
    "isPublicDomain": "true",
    "dateBegin": 1900,
    "dateEnd": 2000,
    "classification": "Paintings",
    "limit": 100,
    "offset": 0
}

# create empty List
all_object_ids = []

# create a loop
while True:
    response = requests.get(url, params=params) # This sends a GET request to The Met API using the URL and parameters you defined.
    response.raise_for_status() # This checks whether the request succeeded.
                                # If the response is successful, Python continues.
                                # If the server returns an HTTP error, such as a 404 or 500, Python raises an exception.
                                # This is helpful because it prevents the program from silently continuing after a failed request.
    data = response.json()      # This converts the API response from JSON into a Python dictionary.

    object_ids = data.get("objectIDs") or [] # This retrieves the value associated with "objectIDs".
    all_object_ids.extend(object_ids) # The .extend() method adds each item from object_ids to the end of all_object_ids.

    print("Retrieved:", len(object_ids), "IDs") # This prints the number of IDs received in the current request.

    params["offset"] += len(object_ids) # This line updates the offset so the next request starts after the IDs already retrieved.
                                        # The += operator means “add to the existing value and save the result.”

    if not object_ids or params["offset"] >= data["total"]: # This line checks whether the loop should stop. 
                                                            # It will stop when no ObjectIDs or when offset count = total count
        break 

print("Total matching objects:", data["total"]) # displays the API's reported total.
print("Total IDs collected:", len(all_object_ids)) # counts the IDs your code actually collected.
print(all_object_ids) # prints the complete list.
```

```python
# Use a loop to request the details for every ID, convert each response into a dictionary, and append it to a new list.

import requests
import time

base_url = "https://collectionapi.metmuseum.org/public/collection/v1/objects/"

all_object_info = [] # creates an empty list to hold the artwork dictionaries.

for object_id in all_object_ids: # loops through every ID you collected in all_object_ids
    url = f"{base_url}{object_id}" # creates the individual object's API URL.

    try:
        response = requests.get(url, timeout=20) # requests the artwork's information.
        response.raise_for_status() # catches HTTP errors.

        object_data = response.json() # converts the JSON response into a Python dictionary.
        all_object_info.append(object_data) # adds that dictionary to your list.

        time.sleep(5) # 5 second pauses briefly between requests.

    except requests.RequestException as error: # so one failed request doesn't stop the whole loop.
        print(f"Could not retrieve object {object_id}: {error}")

print("Number of artworks retrieved:", len(all_object_info))
print(all_object_info[0])
```
