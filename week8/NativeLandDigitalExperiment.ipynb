```python
# first we import everything e need for the task at hand

import requests # this gives us the ability to request data from remote websites
from time import sleep # this gives us the ability to implement time delay sin our code to meet APi regulations
from getpass import getpass # this is a password protection feature for python notebooks

key = getpass("Paste your API key here!")
```

```python
# this method is setting the parameters as variables then slotting them into the URL with {}

maps = "languages" # their date format is in ISO format
coordinates = "42.553080,-86.473389" # coordinates separated by comma

url = f"https://native-land.ca/api/index.php?maps={maps}&position={coordinates}&key={key}"
response = requests.get(url)

print(response.status_code)
print(url)
```

```python
parsed_nativelanddigital_response = response.json() # interprets response as Python-readable JSON
print(parsed_nativelanddigital_response.keys()) # tells me what keys are in the raw response
print(parsed_nativelanddigital_response['response'].keys()) # tells me what keys are inthe "response parent
nativelanddigital_docs= parsed_nativelanddigital_response['response']['docs'] # tells me what keys are in the "docs" key, which is a child of "response"
print(len(nativelanddigital_docs))
```
