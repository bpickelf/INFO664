## cleaning up text with strip ()

```python
messy_string= "       this sentance is full of a bunch of trailing spaces.:(            "
print(messy_string)
print(messy_string.strip())
```

You can find the *length* of a string by using the `len()` function, chich returns the mength of characters

```python
print(len(messy_string))
clean_string=messy_string.strip()
print(len(clean_string))
```

## Text replacement!

`.replace(old, new)` searches for one section of a string and replaces it with another!

```python
mood_string = "I am a terrible time programming in Python"
print(mood_string)
mood_string=mood_string.replace("terrible", "wonderful")
print(mood_string)
```

## parsing comma-deliminated data

python has methods for parsing sata out of strings by delimiting chrracters. 
For instance commas are often separate values in CSV-stype datasets. the `split(",")' method will separate a string based on locations of commas
The example below is goign to separate the data and declare them as four variables at once.

```python
csv_data = "Nirvana, Smells like teen spirit, 5:01, 1991"

artist, song, length, year = csv_data.split(",")

my_string = f"My favorite band is {artist.strip()}, and their best song is {song.strip()}, which was released in {year.strip()} and is {length.strip()} long."

print(my_string)
```

## Slicing a String

We can extract a certain range of characters from a string using the slicing approach:

```python
my_text = "saxaphone"
print(my_text[3]) # shows the 3rd character
print(my_text[5:]) # shows everything after the 5th character
print(my_text[2:5]) # finds characters 2 through 5
print(len(my_text))

print(my_text[-1]) # prints the last character. negative numbers count from right to left.      
print(my_text[-4:])
print(my_text[-4])
print(my_text[-:-4])
```

```python
print(my_text[9]) # this will not work because python counts the first character as ZERO. 9 is the length, but there is no 9th character.
```
