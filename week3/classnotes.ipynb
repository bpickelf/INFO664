## parsing comma-deliminated data

python has methods for parsing sata out of strings by delimiting chrracters. 
For instance commas are often separate values in CSV-stype datasets. the `split(",")' method will separate a string based on locations of commas

```python
csv_data = "Nirvana, Smells like teen spirit, 5:01, 1991"

artist, song, length, year = csv_data.split(",")

my_string = f"My favorite band is {artist.strip()}, and their best song is {song.strip()}, which was released in {year.strip()} and is {length.strip()} long."

print(my_string)
```

```python
print(birds)
print(leap_years)
print(bands)
```

```python
print(birds[0]) #pulls a single value
print(birds[0:3]) #pulls a sub-list of the values from the full list
```

If we want to access a value within a *list of lists* we must use two index nubers:

```python
print(bands[0])  # this gives me the list contained in the list of lists
print(bands[0][-1]) # the -1 indes gives me the laast entry in the first list.

print(bands[2])
print(bands[2][1])
```

## Sorting Dictionaries

when we are sorting lists, we can use a simple sort method that puts items in alphabetical or numerical order. This is more difficult when dealing with *dictionaries* which have more diverse nested data. It's often the case that we don't merely want to alphabetize, but to sort based on a specific nested quality.

```python
print(central_park_birds)
```

```python
# sorted() fuction takes two arguments: one is the dataset to tsort
# the other is the logic by which to sort
# in this case, we use a lambdaa function to specify which key to compare
sorted_central_park_birds=sorted(central_park_birds, key=lambda bird:bird["count"])
print(sorted_central_park_birds)
```

```python
sorted_central_park_birds=sorted(central_park_birds, key=lambda bird:bird["count"], reverse=True) # this optional third argument specifies whether the order should be in descending (reverse) order
print(sorted_central_park_birds)
```

## lambda function

A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

```python
# Add 10 to argument a, and return the result:
# in this example we see a variable also be a function.
x = lambda a : a + 10
print(x(5))
```

```python
# Multiply argument a with argument b and return the result:
x = lambda a, b : a * b
print(x(5, 6))
```
